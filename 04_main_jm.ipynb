```python
def data_generation(num_elements,real_data):

    
    # Choose random data samples to calculate the distance from
    elements1 = np.random.choice(len(real_data), num_elements, replace=True)
    # Choose random data samples to calculate the distance to
    delta = np.random.choice(len(real_data) - 1, num_elements, replace=True) + 1
    elements2 = elements1 + delta
    # Push the second elements to the range if they are outside
    elements2[elements2 >= len(real_data)] = elements2[elements2 >= len(real_data)] - len(real_data)
    distances_L2 = np.linalg.norm(real_data[elements1] - real_data[elements2], axis = 1)
    #print((real_data[elements1]).shape)
    
    # Sort all the distances from the smallest to the largest.
    distances_L2_sorted = np.sort(distances_L2)
    n = len(distances_L2_sorted)

    return distances_L2_sorted,n

def confidence_interval(n,distances_L2_sorted,num_elements,N_val):
    
    # 95 percent confidence intercal
    epsilon = np.sqrt(np.log(2 / 0.05) / (2 * n))  # Using confidence of 95% (alpha = 0.05)
    cdf_lower = np.maximum(0, np.arange(1, n + 1) / n - epsilon)
    cdf_upper = np.minimum(1, np.arange(1, n + 1) / n + epsilon)
    
    # C(r) vs r plot
    x_axis = np.log(distances_L2_sorted)
    y_axis = np.log(range(len(distances_L2_sorted)))
    q = (np.arange(0, len(distances_L2_sorted)) + 0.5)/num_elements
    
    # Slicing the log-log plot to get slopes at different areas
    x_axis = np.log(distances_L2_sorted)
    #N_val = 50
    k = len(y_axis)//N_val
    y_change = y_axis[k::k] - y_axis[0:-k:k]
    x_change = x_axis[k::k] - x_axis[0:-k:k]
    slopes = y_change/x_change
    slopes_number = slopes[~np.isnan(slopes)]
    
    
    # Calculating the confidence intervals
    z_alpha = 2
    up = q + z_alpha * np.sqrt(q*(1-q) / num_elements)
    up[up > 1] = 1
    low = q - z_alpha * np.sqrt(q*(1-q) / num_elements)
    low[low < 0] = 0
    log_up = np.log(up)
    log_low = np.log(low)

    return log_up,log_low,x_axis,q

def estimate_slope(distances_L2_sorted,N_val,log_up,log_low,q):
    

        
    # Calculating the confidence intervals for the interval slices
    dist_ratio = np.exp(np.log(distances_L2_sorted[-1]/distances_L2_sorted[0])/N_val)
    ind1 = 0
    idup = []
    idlow = []
    idest = []
    d = []
    d1=[]
    indices = []
    mean_th=[]
    for i in range(N_val):
        ind2 = np.argmax(distances_L2_sorted > distances_L2_sorted[ind1] * dist_ratio)
        if ind2 == 0 or ind2>=len(distances_L2_sorted):
            break
        else:
            idup.append((log_up[ind2] - log_low[ind1])/(np.log(distances_L2_sorted[ind2]) - np.log(distances_L2_sorted[ind1])) )
            idlow.append((log_low[ind2] - log_up[ind1])/(np.log(distances_L2_sorted[ind2]) - np.log(distances_L2_sorted[ind1])) )
            idest.append((np.log(q[ind2]) - np.log(q[ind1]))/ (np.log(distances_L2_sorted[ind2]) - np.log(distances_L2_sorted[ind1]))) # Estimated slope
            print((np.log(q[ind2]) - np.log(q[ind1])))
            d.append(np.sqrt(distances_L2_sorted[ind2] * distances_L2_sorted[ind1])) # Geometric mean
            indices.append(ind1)
            ind1 = ind2
        #mean_th.append(np.mean(slope_th[(i)*(len(slope_th)//N_val):(i+1)*(len(slope_th)//N_val)]))
    idup = np.array(idup)
    idlow = np.array(idlow)
    idest = np.array(idest)
    #mean_th = np.array(mean_th)
    idlow[idlow < 0] = 0
    indices = np.array(indices)

    return idup,idlow,idest,d
```

```python
# =========================================================
# 1. Configuration
# =========================================================
# Path to the input .npz file
#FILE_PATH = "your_data.npz"

# Dimension of the latent representation
LATENT_DIM = 10

# Training hyperparameters
BATCH_SIZE = 128
EPOCHS = 500
LEARNING_RATE = 1e-3
RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

# =========================================================
# 2. Select computation device
# =========================================================
# Use CUDA if available, otherwise MPS (Apple Silicon),
# otherwise fall back to CPU.
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

print("Using device:", device)

# =========================================================
# 3. Load data from .npz
# =========================================================
# Load the compressed NumPy archive
#npz_data = np.load(FILE_PATH)
# data = np.load("matrices/15000_484_20260414_100939.npz")
#data = np.load("features/20260414_100939_det0000.npz")
data = np.load("matrices/NEW_15000_484_20260414_100939.npz")

data=data['matrix'][:,:484]

#print("Stored arrays:", npz_data.files)

# Use the first array in the archive
#data_key = npz_data.files[0]
X = data

#print("Selected key:", data_key)
print("Original shape:", X.shape)

# Transpose if the data are stored as (484, N)
if X.shape == (484, 15000):
    X = X.T

# Verify that each sample has 484 features
assert X.shape[1] == 484

X = X.astype(np.float32)


# =========================================================
# 4. Normalize features
# =========================================================
# Standardize each feature independently
mean = X.mean(axis=0, keepdims=True)
std = X.std(axis=0, keepdims=True)

# Avoid division by zero
std[std < 1e-8] = 1.0

X = (X - mean) / std

X_tensor = torch.tensor(X, dtype=torch.float32)


# =========================================================
# 5. Split into training and validation sets
# =========================================================
dataset = TensorDataset(X_tensor)

train_size = int(0.8 * len(dataset))
val_size = len(dataset) - train_size

train_dataset, val_dataset = random_split(
    dataset,
    [train_size, val_size],
    generator=torch.Generator().manual_seed(RANDOM_SEED)
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)


# =========================================================
# 6. Define the Autoencoder
# =========================================================
class Autoencoder(nn.Module):
    def __init__(self, input_dim=484, latent_dim=16):
        super().__init__()

        # Encoder: maps input features to a low-dimensional latent space
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.ReLU(),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, latent_dim)
        )

        # Decoder: reconstructs the original input from the latent vector
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 128),
            nn.ReLU(),
            nn.Linear(128, 256),
            nn.ReLU(),
            nn.Linear(256, input_dim)
        )

    def forward(self, x):
        # Encode the input into the latent space
        z = self.encoder(x)

        # Reconstruct the input from the latent representation
        x_hat = self.decoder(z)

        return x_hat


# =========================================================
# 7. Training setup
# =========================================================
model = Autoencoder(
    input_dim=484,
    latent_dim=LATENT_DIM
).to(device)

# Mean squared reconstruction error
criterion = nn.MSELoss()

# Adam optimizer with weight decay
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=1e-5
)


# =========================================================
# 8. Training loop
# =========================================================

train_losses = []
val_losses = []

for epoch in range(EPOCHS):

    model.train()
    train_loss = 0

    for (batch,) in train_loader:

        batch = batch.to(device)

        # Forward pass
        reconstructed = model(batch)

        # Compute reconstruction loss
        loss = criterion(reconstructed, batch)

        # Backpropagation
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        train_loss += loss.item() * batch.size(0)
    
    train_loss /= len(train_loader.dataset)
    train_losses.append(train_loss)


# =========================================================
# 9. Extract latent embeddings
# =========================================================
model.eval()

with torch.no_grad():
    latent_vectors = model.encoder(X_tensor.to(device))
    reconstruction = model(X_tensor.to(device))
    mse = criterion(reconstruction,X_tensor.to(device)).item()

latent_vectors = latent_vectors.cpu().numpy()
print(f"Final reconstruction MSE = {mse:.6f}")
```

```python
# updated autoencoder
"""
484-dim feature matrix -> LATENT_DIM autoencoder.

Changes from the original script:
  * validation set is actually evaluated (val_loader was built but never used)
  * normalization statistics fit on the training split only
  * best-val-loss checkpoint kept instead of whatever epoch 499 produced
  * final MSE reported separately for train and val
  * PCA baseline at the same bottleneck width for comparison
  * per-feature error breakdown
"""

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
from sklearn.decomposition import PCA

# =========================================================
# 1. Configuration
# =========================================================
# FILE_PATH = "features/20260414_100939_det0000.npz"
# FILE_PATH = "matrices/15000_484_20260414_100939.npz"
FILE_PATH = "matrices/NEW_15000_484_20260414_100939.npz"

ARRAY_KEY = "matrix"
N_FEATURES = 484

LATENT_DIM = 10
BATCH_SIZE = 128
EPOCHS = 500
LEARNING_RATE = 1e-3
RANDOM_SEED = 42
TRAIN_FRAC = 0.8

torch.manual_seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

# =========================================================
# 2. Device
# =========================================================
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

print("Using device:", device)

# =========================================================
# 3. Load data
# =========================================================
npz = np.load(FILE_PATH)
X = npz[ARRAY_KEY]
print("Raw shape:", X.shape)

# Orient as (n_samples, n_features) BEFORE slicing columns.
# Slicing [:, :484] first silently mangles the array if it is stored
# transposed as (features, samples).
if X.shape[0] == N_FEATURES and X.shape[1] != N_FEATURES:
    X = X.T
    print("Transposed to:", X.shape)

X = X[:, :N_FEATURES].astype(np.float32)
n_samples = X.shape[0]
print(f"Working shape: {X.shape}  ({n_samples} samples)")

if n_samples < 2000:
    print(
        f"WARNING: only {n_samples} samples for a ~200k-parameter model. "
        "Expect the train/val gap below to be large."
    )

# =========================================================
# 4. Split first, then normalize using training statistics
# =========================================================
rng = np.random.default_rng(RANDOM_SEED)
perm = rng.permutation(n_samples)
n_train = int(TRAIN_FRAC * n_samples)
train_idx, val_idx = perm[:n_train], perm[n_train:]

mean = X[train_idx].mean(axis=0, keepdims=True)
std = X[train_idx].std(axis=0, keepdims=True)

# Track dead/constant features - they inflate "variance explained".
dead = (std < 1e-8).ravel()
if dead.any():
    print(f"NOTE: {dead.sum()} of {N_FEATURES} features are ~constant in train.")
std[std < 1e-8] = 1.0

X = (X - mean) / std
X_tensor = torch.tensor(X, dtype=torch.float32)

train_loader = DataLoader(
    TensorDataset(X_tensor[train_idx]), batch_size=BATCH_SIZE, shuffle=True
)
val_loader = DataLoader(
    TensorDataset(X_tensor[val_idx]), batch_size=BATCH_SIZE, shuffle=False
)

# =========================================================
# 5. Model
# =========================================================
class Autoencoder(nn.Module):
    def __init__(self, input_dim=N_FEATURES, latent_dim=LATENT_DIM):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.ReLU(),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, latent_dim),
        )
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 128),
            nn.ReLU(),
            nn.Linear(128, 256),
            nn.ReLU(),
            nn.Linear(256, input_dim),
        )

    def forward(self, x):
        return self.decoder(self.encoder(x))


model = Autoencoder(input_dim=N_FEATURES, latent_dim=LATENT_DIM).to(device)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(
    model.parameters(), lr=LEARNING_RATE, weight_decay=1e-5
)

# =========================================================
# 6. Training loop with validation
# =========================================================
train_losses, val_losses = [], []
best_val, best_state, best_epoch = float("inf"), None, -1

for epoch in range(EPOCHS):
    model.train()
    train_loss = 0.0
    for (batch,) in train_loader:
        batch = batch.to(device)
        loss = criterion(model(batch), batch)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        train_loss += loss.item() * batch.size(0)
    train_loss /= len(train_loader.dataset)

    model.eval()
    val_loss = 0.0
    with torch.no_grad():
        for (batch,) in val_loader:
            batch = batch.to(device)
            val_loss += criterion(model(batch), batch).item() * batch.size(0)
    val_loss /= len(val_loader.dataset)

    train_losses.append(train_loss)
    val_losses.append(val_loss)

    if val_loss < best_val:
        best_val, best_epoch = val_loss, epoch
        best_state = {
            k: v.detach().cpu().clone() for k, v in model.state_dict().items()
        }

    if epoch % 25 == 0 or epoch == EPOCHS - 1:
        print(f"epoch {epoch:4d}  train {train_loss:.6f}  val {val_loss:.6f}")

model.load_state_dict(best_state)
print(f"\nBest val MSE {best_val:.6f} at epoch {best_epoch} "
      f"(final epoch val was {val_losses[-1]:.6f})")

# =========================================================
# 7. Latents and honest evaluation
# =========================================================
model.eval()
with torch.no_grad():
    Xd = X_tensor.to(device)
    latent_vectors = model.encoder(Xd).cpu().numpy()
    recon = model(Xd).cpu().numpy()


def mse_on(idx):
    return float(np.mean((recon[idx] - X[idx]) ** 2))


train_mse, val_mse = mse_on(train_idx), mse_on(val_idx)

# After z-scoring, predicting the mean gives MSE == 1.0, so
# variance explained is simply 1 - MSE.
pca = PCA(n_components=LATENT_DIM).fit(X[train_idx])
pca_recon = pca.inverse_transform(pca.transform(X[val_idx]))
pca_mse = float(np.mean((pca_recon - X[val_idx]) ** 2))

print("\n--- reconstruction ---")
print(f"train MSE          = {train_mse:.6f}   (var explained {1 - train_mse:.4f})")
print(f"val   MSE          = {val_mse:.6f}   (var explained {1 - val_mse:.4f})")
print(f"PCA-{LATENT_DIM} val MSE     = {pca_mse:.6f}   (var explained {1 - pca_mse:.4f})")
print(f"train/val gap      = {val_mse - train_mse:.6f}")
print(f"AE improvement over PCA = {(pca_mse - val_mse) / pca_mse * 100:.1f}%")

# Which features are hard to reconstruct?
per_feat = np.mean((recon[val_idx] - X[val_idx]) ** 2, axis=0)
order = np.argsort(per_feat)
print("\nbest-reconstructed features:", order[:5].tolist(),
      np.round(per_feat[order[:5]], 4).tolist())
print("worst-reconstructed features:", order[-5:].tolist(),
      np.round(per_feat[order[-5:]], 4).tolist())

print(f"\nlatent_vectors shape: {latent_vectors.shape}")
```

```python
"""
Detect-then-encode: bat call patches -> 4-dimensional code -> reconstruction.

This is the configuration that reached 66.7% of call variance (79% of
recoverable variance) with 4 dimensions, against 17.7% and 23.6% for
autoencoders over continuous spectrogram frames.

Self-contained: detects calls, cuts patches, runs leave-one-call-out, and
compares against PCA at the same width. That last comparison is the new part -
on continuous frames the autoencoder tied PCA at both 20 and 30 dimensions, so
the question is whether the nonlinearity earns its place HERE.

Why this representation works where the continuous one did not:
  * 1.33 ms frames instead of 40 ms, so a 2 ms chirp is not averaged away
  * 45-100 kHz, above the insect chorus
  * one training row per detected call, not per frame of mostly-empty night
  * 2D patches over time x frequency, so a sweep is a diagonal ridge the
    encoder can represent
"""

import os

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import soundfile as sf
import matplotlib.pyplot as plt
from torch.utils.data import TensorDataset, DataLoader
from sklearn.decomposition import PCA

# =========================================================
# Configuration
# =========================================================
FULL_WAV = "/home/jgersey/SageBat/20260414_100939.wav"
OUT = "autoencoder_results"

# spectrogram - sized for millisecond transients
N_FFT, HOP = 512, 128
F_LO, F_HI = 45_000, 100_000

# detection
PCTL, THRESH_DB, MIN_GAP = 90, 5, 3
PATCH_T = 24

# augmentation (training folds only)
SHIFTS = [-3, 0, 3]
GAINS_DB = [-2, 0, 2]

# model
LATENT = 4
BATCH_SIZE = 64
EPOCHS = 300
LR = 1e-3
WEIGHT_DECAY = 1e-4
PATIENCE = 40
SEED = 42

torch.manual_seed(SEED)
np.random.seed(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device)
os.makedirs(OUT, exist_ok=True)


# =========================================================
# 1. Detect calls
# =========================================================
info = sf.info(FULL_WAV)
SR = info.samplerate
WIN = np.hanning(N_FFT)
FREQS = np.fft.rfftfreq(N_FFT, d=1 / SR)
KEEP = np.where((FREQS >= F_LO) & (FREQS <= F_HI))[0]
BAND = FREQS[KEEP]
N_BINS = len(KEEP)

print(f"{info.duration:.0f} s at {SR} Hz")
print(f"frame {N_FFT/SR*1000:.2f} ms, hop {HOP/SR*1000:.2f} ms, {N_BINS} bins "
      f"({BAND[0]/1000:.0f}-{BAND[-1]/1000:.0f} kHz)")


def stft_db(x):
    n = 1 + (len(x) - N_FFT) // HOP
    fr = np.lib.stride_tricks.as_strided(
        x, shape=(n, N_FFT), strides=(x.strides[0] * HOP, x.strides[0]))
    p = (np.abs(np.fft.rfft(fr * WIN, axis=1)) ** 2)[:, KEEP]
    return 10 * np.log10(np.maximum(p, 1e-20))


times = []
BLK = int(10.0 * SR)
with sf.SoundFile(FULL_WAV) as f:
    b = 0
    while True:
        x = f.read(frames=BLK, dtype="float64", always_2d=True)
        if len(x) < N_FFT * 4:
            break
        x = np.ascontiguousarray(x[:, 0])
        d = stft_db(x)
        e = np.percentile(d, PCTL, axis=1)
        hits = np.where(e > np.median(e) + THRESH_DB)[0]
        if len(hits):
            for g in np.split(hits, np.where(np.diff(hits) > MIN_GAP)[0] + 1):
                times.append(b * 10.0 + int(g[np.argmax(e[g])]) * HOP / SR)
        b += 1

times = np.array(sorted(times))
print(f"\n{len(times)} calls detected")
if len(times) < 5:
    raise SystemExit("too few calls for leave-one-out")


# =========================================================
# 2. Cut patches
# =========================================================
CONTEXT_S = 0.4


def patch_at(t, shift=0):
    with sf.SoundFile(FULL_WAV) as f:
        f.seek(max(0, int((t - CONTEXT_S / 2) * SR)))
        x = np.ascontiguousarray(
            f.read(frames=int(CONTEXT_S * SR), dtype="float64", always_2d=True)[:, 0])
    if len(x) < N_FFT * 4:
        return None
    d = stft_db(x)
    c = d.shape[0] // 2 + shift
    lo, hi = c - PATCH_T // 2, c + PATCH_T // 2
    return d[lo:hi] if lo >= 0 and hi <= d.shape[0] else None


calls, keep_t = [], []
for t in times:
    p = patch_at(t)
    if p is not None:
        calls.append(p)
        keep_t.append(t)
calls = np.stack(calls).astype(np.float32)
keep_t = np.array(keep_t)
N = len(calls)
print(f"{N} patches of {PATCH_T}x{N_BINS} ({PATCH_T*N_BINS} values each)")

# background controls from quiet stretches
gaps = []
for i in range(len(times) - 1):
    if times[i + 1] - times[i] > 2.0:
        gaps.append((times[i] + 0.7, times[i + 1] - 0.7))
rng = np.random.default_rng(SEED)
bg = []
while len(bg) < N * 3 and gaps:
    lo, hi = gaps[rng.integers(len(gaps))]
    if hi > lo:
        p = patch_at(float(rng.uniform(lo, hi)))
        if p is not None:
            bg.append(p)
BG = np.stack(bg).astype(np.float32) if bg else None
print(f"{0 if BG is None else len(BG)} background control patches")


# =========================================================
# 3. Model
# =========================================================
class TinyConvAE(nn.Module):
    def __init__(self, t, f, latent, ch=(8, 16)):
        super().__init__()
        c1, c2 = ch
        self.t, self.f = t, f
        self.enc = nn.Sequential(
            nn.Conv2d(1, c1, 3, stride=2, padding=1), nn.ReLU(),
            nn.Conv2d(c1, c2, 3, stride=2, padding=1), nn.ReLU())
        with torch.no_grad():
            shp = self.enc(torch.zeros(1, 1, t, f)).shape[1:]
        self.shp = shp
        flat = int(np.prod(shp))
        self.fc_e = nn.Linear(flat, latent)
        self.fc_d = nn.Linear(latent, flat)
        self.dec = nn.Sequential(
            nn.ConvTranspose2d(c2, c1, 3, stride=2, padding=1, output_padding=1), nn.ReLU(),
            nn.ConvTranspose2d(c1, c1, 3, stride=2, padding=1, output_padding=1), nn.ReLU(),
            nn.Conv2d(c1, 1, 3, padding=1))

    def encode(self, x):
        return self.fc_e(self.enc(x.unsqueeze(1)).flatten(1))

    def forward(self, x):
        h = self.fc_d(self.encode(x)).view(-1, *self.shp)
        o = self.dec(h)
        return F.interpolate(o, size=(self.t, self.f), mode="bilinear",
                             align_corners=False).squeeze(1)


crit = nn.MSELoss()


# =========================================================
# 4. Leave-one-call-out, AE and PCA
# =========================================================
def run_fold(held, latent=LATENT):
    tr = []
    for i in range(N):
        if i == held:
            continue
        for s in SHIFTS:
            p = patch_at(keep_t[i], s)
            if p is not None:
                tr.append(p)
    tr = np.stack(tr).astype(np.float32)
    tr = np.concatenate([tr + g for g in GAINS_DB], axis=0)
    va = calls[held:held + 1]

    mu = tr.mean(axis=(0, 1), keepdims=True)
    sd = tr.std(axis=(0, 1), keepdims=True)
    sd[sd < 1e-8] = 1.0
    trs, vas = (tr - mu) / sd, (va - mu) / sd
    bgs = (BG - mu) / sd if BG is not None else None

    # --- autoencoder ---
    loader = DataLoader(TensorDataset(torch.tensor(trs)),
                        batch_size=BATCH_SIZE, shuffle=True)
    m = TinyConvAE(PATCH_T, N_BINS, latent).to(device)
    opt = torch.optim.Adam(m.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
    vt = torch.tensor(vas).to(device)
    best, state, best_ep = float("inf"), None, -1
    for ep in range(EPOCHS):
        m.train()
        for (b,) in loader:
            b = b.to(device)
            loss = crit(m(b), b)
            opt.zero_grad(); loss.backward(); opt.step()
        m.eval()
        with torch.no_grad():
            v = crit(m(vt), vt).item()
        if v < best:
            best, best_ep = v, ep
            state = {k: t.detach().cpu().clone() for k, t in m.state_dict().items()}
        if ep - best_ep > PATIENCE:
            break
    m.load_state_dict(state)
    m.eval()
    with torch.no_grad():
        rec = m(vt).cpu().numpy()[0]
        bg_mse = (crit(m(torch.tensor(bgs).to(device)),
                       torch.tensor(bgs).to(device)).item()
                  if bgs is not None else np.nan)

    # --- PCA at the same width, same data ---
    trf = trs.reshape(len(trs), -1)
    vaf = vas.reshape(1, -1)
    k = min(latent, len(trf) - 1, trf.shape[1])
    pca = PCA(n_components=k).fit(trf)
    prec = pca.inverse_transform(pca.transform(vaf)).reshape(PATCH_T, N_BINS)

    return dict(held=held, orig=vas[0], ae=rec, pca=prec,
                ae_mse=float(((rec - vas[0]) ** 2).mean()),
                pca_mse=float(((prec - vas[0]) ** 2).mean()),
                bg_mse=bg_mse)


print(f"\nleave-one-call-out, latent {LATENT} ...")
folds = []
for h in range(N):
    r = run_fold(h)
    folds.append(r)
    print(f"  call {h:2d} @ {keep_t[h]:7.2f}s   AE {r['ae_mse']:.3f}   "
          f"PCA {r['pca_mse']:.3f}   bg {r['bg_mse']:.3f}")


# =========================================================
# 5. Call-region metrics
# =========================================================
def region_stats(key):
    ce, cb, be = [], [], []
    for r in folds:
        o = r["orig"]
        mask = o > o.mean() + 1.5 * o.std()
        if mask.sum() < 5:
            continue
        e = (r[key] - o) ** 2
        ce.append(float(e[mask].mean()))
        cb.append(float((o[mask] ** 2).mean()))   # cost of predicting zero
        be.append(float(e[~mask].mean()))
    return np.array(ce), np.array(cb), np.array(be)


ae_ce, base, ae_be = region_stats("ae")
pca_ce, _, pca_be = region_stats("pca")

ae_ve = 1 - ae_ce.mean() / base.mean()
pca_ve = 1 - pca_ce.mean() / base.mean()
ae_ve_corr = 1 - (ae_ce.mean() - ae_be.mean()) / (base.mean() - ae_be.mean())

ae_all = np.array([r["ae_mse"] for r in folds])
pca_all = np.array([r["pca_mse"] for r in folds])
bg_all = np.array([r["bg_mse"] for r in folds])

print("\n" + "=" * 62)
print(f"{N} calls, leave-one-out, latent {LATENT}")
print(f"  input {PATCH_T}x{N_BINS} = {PATCH_T*N_BINS} values -> {LATENT} "
      f"({PATCH_T*N_BINS/LATENT:.0f}x compression)")
print("\n--- whole patch ---")
print(f"  AE  MSE   {ae_all.mean():.4f} +/- {ae_all.std():.4f}")
print(f"  PCA MSE   {pca_all.mean():.4f} +/- {pca_all.std():.4f}")
print(f"  background {bg_all.mean():.4f}   (per-pixel noise floor)")
print(f"  AE vs PCA {(pca_all.mean()-ae_all.mean())/pca_all.mean()*100:+.1f}%")

print("\n--- call region only (the number that matters) ---")
print(f"  baseline (predict zero)   {base.mean():.3f}")
print(f"  AE  call-pixel MSE        {ae_ce.mean():.3f}   -> {ae_ve*100:.1f}% explained")
print(f"  PCA call-pixel MSE        {pca_ce.mean():.3f}   -> {pca_ve*100:.1f}% explained")
print(f"  AE, noise-floor corrected -> {ae_ve_corr*100:.1f}% of recoverable")
gain = (pca_ce.mean() - ae_ce.mean()) / pca_ce.mean() * 100
print(f"  AE vs PCA on call pixels  {gain:+.1f}%")

print("\n--- comparison with continuous-frame autoencoders ---")
print(f"  1-second matrix,  484->20 : 17.7% var explained, ties PCA")
print(f"  10-minute matrix, 484->30 : 23.6% var explained, ties PCA")
print(f"  call patches,   {PATCH_T*N_BINS}->{LATENT:<2d} : {ae_ve*100:.1f}% of call variance, "
      f"{gain:+.0f}% vs PCA")
if gain > 10:
    print("\n  -> the nonlinearity earns its place on patches, unlike on frames.")
elif gain > 0:
    print("\n  -> small nonlinear gain; report AE and PCA side by side.")
else:
    print("\n  -> PCA matches the AE here too: the representation, not the")
    print("     model, is what makes this work. Still a valid and simpler result.")


# =========================================================
# 6. Figures
# =========================================================
ext = [0, PATCH_T * HOP / SR * 1000, BAND[0] / 1000, BAND[-1] / 1000]
sel = np.argsort([r["ae_mse"] for r in folds])[[0, N // 3, 2 * N // 3, -1]]

fig, axes = plt.subplots(4, len(sel), figsize=(3.4 * len(sel), 10), squeeze=False)
for j, i in enumerate(sel):
    r = folds[i]
    for row, (data, lab, cmap) in enumerate([
            (r["orig"], "original", "inferno"),
            (r["ae"], f"AE ({LATENT} dims)", "inferno"),
            (r["pca"], f"PCA-{LATENT}", "inferno"),
            (r["orig"] - r["ae"], "AE residual", "RdBu_r")]):
        a = axes[row][j]
        if cmap == "RdBu_r":
            v = np.abs(data).max()
            a.imshow(data.T, aspect="auto", origin="lower", cmap=cmap,
                     extent=ext, vmin=-v, vmax=v)
        else:
            a.imshow(data.T, aspect="auto", origin="lower", cmap=cmap, extent=ext)
        if j == 0:
            a.set_ylabel(lab, fontsize=9)
        a.tick_params(labelsize=7)
    axes[0][j].set_title(f"call @ {keep_t[i]:.2f}s\nAE {r['ae_mse']:.2f}", fontsize=8)
fig.supxlabel("ms"); fig.supylabel("kHz")
plt.suptitle(f"Detected call patches: {PATCH_T*N_BINS} values -> {LATENT} dimensions",
             y=1.00)
plt.tight_layout()
plt.savefig(os.path.join(OUT, "patch_ae_vs_pca.png"), dpi=140, bbox_inches="tight")
plt.show()

fig, ax = plt.subplots(1, 2, figsize=(13, 4.2))
x = np.arange(N)
ax[0].bar(x - 0.2, ae_all, 0.4, label="AE")
ax[0].bar(x + 0.2, pca_all, 0.4, label=f"PCA-{LATENT}")
ax[0].axhline(bg_all.mean(), ls="--", c="k", lw=0.9, label="background")
ax[0].set_xlabel("held-out call"); ax[0].set_ylabel("MSE")
ax[0].legend(fontsize=8); ax[0].grid(alpha=0.3, axis="y")
ax[0].set_title("per-fold reconstruction error")

labels = ["1 s\n484→20", "10 min\n484→30", f"patches\n{PATCH_T*N_BINS}→{LATENT}"]
vals = [0.177, 0.236, ae_ve]
bars = ax[1].bar(labels, vals, color=["tab:gray", "tab:gray", "tab:green"])
for b, v in zip(bars, vals):
    ax[1].text(b.get_x() + b.get_width() / 2, v, f"{v*100:.0f}%",
               ha="center", va="bottom", fontsize=11)
ax[1].set_ylabel("variance explained"); ax[1].set_ylim(0, 1.0)
ax[1].set_title("continuous frames vs detected calls")
ax[1].grid(alpha=0.3, axis="y")
plt.tight_layout()
plt.savefig(os.path.join(OUT, "patch_vs_continuous.png"), dpi=140, bbox_inches="tight")
plt.show()

np.savez_compressed(os.path.join(OUT, "patch_results.npz"),
                    ae_mse=ae_all, pca_mse=pca_all, bg_mse=bg_all,
                    call_times=keep_t, latent=LATENT)
print(f"\nwrote figures and results to {OUT}/")
```

```python
"""
Next tests for the detect-then-encode result.

Ordered by what each would change:

  A. LATENT SWEEP - 4 dimensions was a small-data constraint, not an optimum.
     If the elbow is at 8, the headline compression number changes.

  B. BOUT-LEVEL TRANSFER - leave-one-call-out holds out one call from a bat
     whose other calls are in training. Training on bout 0 and testing on
     bout 1 asks whether the code transfers to an UNSEEN INDIVIDUAL, which is
     what the edge architecture actually requires.

  C. LATENT INTERPRETABILITY - do the dimensions track peak frequency,
     duration, bandwidth, sweep slope? "4 numbers encode the physical call
     parameters" is a far stronger claim than any reconstruction score, and it
     is what makes the code useful for downstream classification.

  D. BIT BUDGET - the architecture claim is about transmission. How many bits
     per dimension survive before reconstruction degrades, and how does that
     compare against simply compressing the patch?

Run after patch_autoencoder_run.py, which defines: calls, keep_t, N, PATCH_T,
N_BINS, BAND, HOP, SR, run_fold, TinyConvAE, device, SHIFTS, GAINS_DB,
patch_at, BG, LATENT, OUT.
"""

import os
import numpy as np
import torch
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.linear_model import LinearRegression

BOUT_SPLIT_S = 400.0     # calls before this are bout 0, after are bout 1
QUANT_BITS = [16, 8, 6, 4, 3, 2]


def call_region_ve(recons, origs):
    """Variance explained on call pixels, predict-zero baseline."""
    ce, cb, be = [], [], []
    for r, o in zip(recons, origs):
        m = o > o.mean() + 1.5 * o.std()
        if m.sum() < 5:
            continue
        e = (r - o) ** 2
        ce.append(e[m].mean()); cb.append((o[m] ** 2).mean()); be.append(e[~m].mean())
    ce, cb, be = np.mean(ce), np.mean(cb), np.mean(be)
    return 1 - ce / cb, 1 - (ce - be) / (cb - be), ce, be


# =========================================================
# A. Latent sweep
# =========================================================
def test_a(widths=(2, 3, 4, 6, 8, 12, 16, 24)):
    print("\n" + "=" * 62)
    print("A. LATENT SWEEP (leave-one-call-out, call-region metric)")
    print("=" * 62)
    rows = []
    for d in widths:
        folds = [run_fold(h, latent=d) for h in range(N)]
        ae_ve, ae_corr, ae_ce, ae_be = call_region_ve(
            [f["ae"] for f in folds], [f["orig"] for f in folds])
        pca_ve, _, pca_ce, _ = call_region_ve(
            [f["pca"] for f in folds], [f["orig"] for f in folds])
        gain = (pca_ce - ae_ce) / pca_ce * 100
        rows.append((d, ae_ve, pca_ve, gain, ae_corr))
        print(f"  latent {d:2d}:  AE {ae_ve*100:5.1f}%   PCA {pca_ve*100:5.1f}%   "
              f"AE-PCA {gain:+5.1f}%   (recoverable {ae_corr*100:5.1f}%)")

    d, ae, pca, gain, corr = zip(*rows)
    fig, ax = plt.subplots(1, 2, figsize=(13, 4))
    ax[0].plot(d, np.array(ae) * 100, "o-", label="autoencoder")
    ax[0].plot(d, np.array(pca) * 100, "s--", label="PCA")
    ax[0].set_xlabel("latent dimensions"); ax[0].set_ylabel("% call variance")
    ax[0].legend(); ax[0].grid(alpha=0.3); ax[0].set_title("look for the elbow")
    ax[1].axhline(0, c="k", lw=0.8)
    ax[1].plot(d, gain, "o-", color="tab:purple")
    ax[1].set_xlabel("latent dimensions"); ax[1].set_ylabel("AE advantage over PCA (%)")
    ax[1].grid(alpha=0.3); ax[1].set_title("does nonlinearity ever pay?")
    plt.tight_layout()
    plt.savefig(os.path.join(OUT, "test_a_latent_sweep.png"), dpi=140,
                bbox_inches="tight")
    plt.show()
    return rows


# =========================================================
# B. Bout-level transfer
# =========================================================
def test_b(latent=None):
    latent = latent or LATENT
    print("\n" + "=" * 62)
    print("B. BOUT TRANSFER - train on bout 0, test on the unseen bat in bout 1")
    print("=" * 62)

    b0 = np.where(keep_t < BOUT_SPLIT_S)[0]
    b1 = np.where(keep_t >= BOUT_SPLIT_S)[0]
    print(f"  bout 0: {len(b0)} calls    bout 1: {len(b1)} calls")
    if len(b1) < 2:
        print("  only "
              f"{len(b1)} call(s) in bout 1 - result will be very noisy")

    tr = []
    for i in b0:
        for s in SHIFTS:
            p = patch_at(keep_t[i], s)
            if p is not None:
                tr.append(p)
    tr = np.stack(tr).astype(np.float32)
    tr = np.concatenate([tr + g for g in GAINS_DB], axis=0)
    va = calls[b1]

    mu = tr.mean(axis=(0, 1), keepdims=True)
    sd = tr.std(axis=(0, 1), keepdims=True); sd[sd < 1e-8] = 1.0
    trs, vas = (tr - mu) / sd, (va - mu) / sd

    from torch.utils.data import TensorDataset, DataLoader
    import torch.nn as nn
    loader = DataLoader(TensorDataset(torch.tensor(trs)), batch_size=64, shuffle=True)
    m = TinyConvAE(PATCH_T, N_BINS, latent).to(device)
    opt = torch.optim.Adam(m.parameters(), lr=1e-3, weight_decay=1e-4)
    crit = nn.MSELoss()
    vt = torch.tensor(vas).to(device)
    best, state, best_ep = float("inf"), None, -1
    for ep in range(300):
        m.train()
        for (b,) in loader:
            b = b.to(device)
            loss = crit(m(b), b)
            opt.zero_grad(); loss.backward(); opt.step()
        m.eval()
        with torch.no_grad():
            v = crit(m(vt), vt).item()
        if v < best:
            best, best_ep, state = v, ep, {k: t.detach().cpu().clone()
                                           for k, t in m.state_dict().items()}
        if ep - best_ep > 40:
            break
    m.load_state_dict(state); m.eval()
    with torch.no_grad():
        rec = m(vt).cpu().numpy()

    trf = trs.reshape(len(trs), -1)
    p = PCA(n_components=min(latent, len(trf) - 1)).fit(trf)
    prec = p.inverse_transform(p.transform(vas.reshape(len(vas), -1)))
    prec = prec.reshape(-1, PATCH_T, N_BINS)

    ae_ve, ae_corr, _, _ = call_region_ve(list(rec), list(vas))
    pca_ve, _, _, _ = call_region_ve(list(prec), list(vas))

    print(f"\n  held-out bat, AE  : {ae_ve*100:.1f}% of call variance")
    print(f"  held-out bat, PCA : {pca_ve*100:.1f}%")
    print("\n  Compare against the leave-one-call-out number. A large drop means")
    print("  the code was fitting one individual, not bat calls generally -")
    print("  which is the thing the edge architecture needs it not to do.")

    n = min(3, len(vas))
    ext = [0, PATCH_T * HOP / SR * 1000, BAND[0] / 1000, BAND[-1] / 1000]
    fig, ax = plt.subplots(3, n, figsize=(3.4 * n, 8), squeeze=False)
    for j in range(n):
        for row, (dta, lab, cm) in enumerate([
                (vas[j], "original (unseen bat)", "inferno"),
                (rec[j], f"AE ({latent} dims)", "inferno"),
                (vas[j] - rec[j], "residual", "RdBu_r")]):
            a = ax[row][j]
            if cm == "RdBu_r":
                v = np.abs(dta).max()
                a.imshow(dta.T, aspect="auto", origin="lower", cmap=cm,
                         extent=ext, vmin=-v, vmax=v)
            else:
                a.imshow(dta.T, aspect="auto", origin="lower", cmap=cm, extent=ext)
            if j == 0:
                a.set_ylabel(lab, fontsize=9)
    plt.suptitle("Bout 1 reconstructed by a model that never saw this bat", y=1.0)
    plt.tight_layout()
    plt.savefig(os.path.join(OUT, "test_b_bout_transfer.png"), dpi=140,
                bbox_inches="tight")
    plt.show()
    return ae_ve, pca_ve


# =========================================================
# C. Do the latents encode call parameters?
# =========================================================
def measure(patch):
    """Physical parameters of one call patch."""
    z = (patch - patch.mean()) / (patch.std() + 1e-9)
    mask = z > 1.5
    if mask.sum() < 5:
        return None
    frames = np.where(mask.any(axis=1))[0]
    bins = np.where(mask.any(axis=0))[0]
    peak_f = BAND[int(z.max(axis=0).argmax())] / 1000
    trace = np.array([BAND[int(z[t].argmax())] / 1000 for t in frames])
    return dict(
        peak_khz=peak_f,
        start_khz=trace[0],
        end_khz=trace[-1],
        slope=(trace[-1] - trace[0]) / max(len(frames), 1),
        duration_ms=len(frames) * HOP / SR * 1000,
        bandwidth_khz=(BAND[bins.max()] - BAND[bins.min()]) / 1000,
        energy=float(patch[mask].mean()),
    )


def test_c(latent=None):
    latent = latent or LATENT
    print("\n" + "=" * 62)
    print("C. DO THE LATENT DIMENSIONS ENCODE CALL PARAMETERS?")
    print("=" * 62)

    mu = calls.mean(axis=(0, 1), keepdims=True)
    sd = calls.std(axis=(0, 1), keepdims=True); sd[sd < 1e-8] = 1.0
    Xs = ((calls - mu) / sd).astype(np.float32)

    # a single model over all calls - this is a descriptive test, not held out
    from torch.utils.data import TensorDataset, DataLoader
    import torch.nn as nn
    aug = np.concatenate([Xs + g / sd.mean() for g in GAINS_DB], axis=0)
    loader = DataLoader(TensorDataset(torch.tensor(aug)), batch_size=64, shuffle=True)
    m = TinyConvAE(PATCH_T, N_BINS, latent).to(device)
    opt = torch.optim.Adam(m.parameters(), lr=1e-3, weight_decay=1e-4)
    crit = nn.MSELoss()
    for ep in range(300):
        m.train()
        for (b,) in loader:
            b = b.to(device)
            loss = crit(m(b), b)
            opt.zero_grad(); loss.backward(); opt.step()
    m.eval()
    with torch.no_grad():
        Z = m.encode(torch.tensor(Xs).to(device)).cpu().numpy()

    params, idx = [], []
    for i in range(N):
        p = measure(calls[i])
        if p:
            params.append(p); idx.append(i)
    if len(params) < 6:
        print(f"  only {len(params)} calls measurable - too few")
        return None
    keys = list(params[0])
    P = np.array([[p[k] for k in keys] for p in params])
    Zi = Z[idx]

    print(f"\n  {len(params)} calls measured")
    print(f"\n  {'parameter':16s}" + "".join(f"  z{i}  " for i in range(latent))
          + "   R2(all)")
    for j, k in enumerate(keys):
        y = P[:, j]
        if np.std(y) < 1e-9:
            continue
        cors = [np.corrcoef(Zi[:, d], y)[0, 1] for d in range(latent)]
        r2 = LinearRegression().fit(Zi, y).score(Zi, y)
        row = "".join(f"{c:+5.2f} " for c in cors)
        flag = "  <-- encoded" if r2 > 0.5 else ""
        print(f"  {k:16s}" + row + f"   {r2:5.2f}{flag}")

    print("\n  R2 above ~0.5 means the latent code carries that physical")
    print("  parameter. Those are the dimensions worth naming on a slide, and")
    print("  they are what would let a server classify from the code alone.")
    print("  Caveat: fitted on all calls, so this is descriptive, not held out.")

    fig, ax = plt.subplots(1, min(3, len(keys)), figsize=(13, 3.8))
    for a, k in zip(np.atleast_1d(ax), keys[:3]):
        y = P[:, keys.index(k)]
        d = int(np.argmax([abs(np.corrcoef(Zi[:, i], y)[0, 1]) for i in range(latent)]))
        a.scatter(Zi[:, d], y, s=40)
        a.set_xlabel(f"latent dim {d}"); a.set_ylabel(k)
        a.set_title(f"r = {np.corrcoef(Zi[:, d], y)[0,1]:+.2f}")
        a.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(OUT, "test_c_latent_meaning.png"), dpi=140,
                bbox_inches="tight")
    plt.show()
    return Z, P, keys


# =========================================================
# D. Bit budget
# =========================================================
def test_d(latent=None):
    latent = latent or LATENT
    print("\n" + "=" * 62)
    print("D. BIT BUDGET - what actually has to be transmitted")
    print("=" * 62)

    import zlib
    folds = [run_fold(h, latent=latent) for h in range(N)]
    base_ve, _, _, _ = call_region_ve([f["ae"] for f in folds],
                                      [f["orig"] for f in folds])

    mu = calls.mean(axis=(0, 1), keepdims=True)
    sd = calls.std(axis=(0, 1), keepdims=True); sd[sd < 1e-8] = 1.0
    Xs = ((calls - mu) / sd).astype(np.float32)

    from torch.utils.data import TensorDataset, DataLoader
    import torch.nn as nn
    loader = DataLoader(TensorDataset(torch.tensor(Xs)), batch_size=64, shuffle=True)
    m = TinyConvAE(PATCH_T, N_BINS, latent).to(device)
    opt = torch.optim.Adam(m.parameters(), lr=1e-3, weight_decay=1e-4)
    crit = nn.MSELoss()
    for ep in range(300):
        m.train()
        for (b,) in loader:
            b = b.to(device)
            loss = crit(m(b), b)
            opt.zero_grad(); loss.backward(); opt.step()
    m.eval()
    with torch.no_grad():
        Z = m.encode(torch.tensor(Xs).to(device)).cpu().numpy()

    lo, hi = Z.min(axis=0), Z.max(axis=0)
    print(f"\n  {'bits/dim':>9s} {'bytes/call':>11s} {'vs raw patch':>13s} "
          f"{'call var expl':>14s}")
    results = []
    for bits in QUANT_BITS:
        levels = 2 ** bits - 1
        Zq = np.round((Z - lo) / (hi - lo + 1e-12) * levels) / levels
        Zq = Zq * (hi - lo) + lo
        with torch.no_grad():
            rec = m.decoder(torch.tensor(Zq, dtype=torch.float32).to(device)) \
                if hasattr(m, "decoder") else None
        if rec is None:
            with torch.no_grad():
                h = m.fc_d(torch.tensor(Zq, dtype=torch.float32).to(device))
                h = h.view(-1, *m.shp)
                import torch.nn.functional as Fn
                o = m.dec(h)
                rec = Fn.interpolate(o, size=(PATCH_T, N_BINS), mode="bilinear",
                                     align_corners=False).squeeze(1)
        rec = rec.cpu().numpy()
        ve, _, _, _ = call_region_ve(list(rec), list(Xs))
        nbytes = latent * bits / 8
        results.append((bits, nbytes, ve))
        print(f"  {bits:9d} {nbytes:11.1f} {PATCH_T*N_BINS/nbytes:12.0f}x "
              f"{ve*100:13.1f}%")

    raw_f32 = PATCH_T * N_BINS * 4
    raw_i8 = PATCH_T * N_BINS
    q = np.clip((Xs - Xs.min()) / (Xs.max() - Xs.min()) * 255, 0, 255).astype(np.uint8)
    zl = np.mean([len(zlib.compress(q[i].tobytes(), 6)) for i in range(N)])

    print(f"\n  baselines per call:")
    print(f"    raw patch float32      {raw_f32:8.0f} bytes")
    print(f"    raw patch int8         {raw_i8:8.0f} bytes")
    print(f"    int8 + zlib            {zl:8.0f} bytes")
    print(f"    {latent} latents @ 8 bit      {latent:8.0f} bytes")
    print(f"\n  vs the honest baseline (int8+zlib): "
          f"{zl/(latent):.0f}x smaller")
    print(f"  1 s of raw audio at 384 kHz/16-bit = {384000*2/1024:.0f} KB, so")
    print(f"  detection alone already provides most of the reduction.")

    bits_, nb, ve = zip(*results)
    plt.figure(figsize=(7, 4))
    plt.plot(bits_, np.array(ve) * 100, "o-")
    plt.axhline(base_ve * 100, ls="--", c="gray", label="float32 LOO")
    plt.xlabel("bits per latent dimension"); plt.ylabel("% call variance")
    plt.title("how much precision actually has to be sent")
    plt.legend(); plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(OUT, "test_d_bit_budget.png"), dpi=140,
                bbox_inches="tight")
    plt.show()
    return results


if __name__ == "__main__":
    print(__doc__)
```

```python
# =========================================================
# Run all four follow-up tests
# Paste AFTER patch_autoencoder_run.py has finished in this kernel.
# =========================================================
import os, zlib
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as Fn
import matplotlib.pyplot as plt
from torch.utils.data import TensorDataset, DataLoader
from sklearn.decomposition import PCA
from sklearn.linear_model import LinearRegression

BOUT_SPLIT_S = 400.0
QUANT_BITS   = [16, 8, 6, 4, 3, 2]
crit = nn.MSELoss()


def call_region_ve(recons, origs):
    ce, cb, be = [], [], []
    for r, o in zip(recons, origs):
        m = o > o.mean() + 1.5 * o.std()
        if m.sum() < 5:
            continue
        e = (r - o) ** 2
        ce.append(e[m].mean()); cb.append((o[m] ** 2).mean()); be.append(e[~m].mean())
    ce, cb, be = np.mean(ce), np.mean(cb), np.mean(be)
    return 1 - ce / cb, 1 - (ce - be) / (cb - be), ce, be


def train_ae(patches_std, latent, epochs=300):
    loader = DataLoader(TensorDataset(torch.tensor(patches_std)),
                        batch_size=64, shuffle=True)
    m = TinyConvAE(PATCH_T, N_BINS, latent).to(device)
    opt = torch.optim.Adam(m.parameters(), lr=1e-3, weight_decay=1e-4)
    for _ in range(epochs):
        m.train()
        for (b,) in loader:
            b = b.to(device)
            loss = crit(m(b), b)
            opt.zero_grad(); loss.backward(); opt.step()
    m.eval()
    return m


def decode(m, Z):
    with torch.no_grad():
        h = m.fc_d(torch.tensor(Z, dtype=torch.float32).to(device)).view(-1, *m.shp)
        o = m.dec(h)
        return Fn.interpolate(o, size=(PATCH_T, N_BINS), mode="bilinear",
                              align_corners=False).squeeze(1).cpu().numpy()


ext = [0, PATCH_T * HOP / SR * 1000, BAND[0] / 1000, BAND[-1] / 1000]

# =========================================================
# B first — it changes what you can claim
# =========================================================
print("=" * 62); print("B. BOUT TRANSFER — unseen individual"); print("=" * 62)

b0 = np.where(keep_t < BOUT_SPLIT_S)[0]
b1 = np.where(keep_t >= BOUT_SPLIT_S)[0]
print(f"  bout 0: {len(b0)} calls    bout 1: {len(b1)} calls")

tr = np.stack([p for i in b0 for s in SHIFTS
               if (p := patch_at(keep_t[i], s)) is not None]).astype(np.float32)
tr = np.concatenate([tr + g for g in GAINS_DB], axis=0)
va = calls[b1]

mu = tr.mean(axis=(0, 1), keepdims=True)
sd = tr.std(axis=(0, 1), keepdims=True); sd[sd < 1e-8] = 1.0
trs, vas = (tr - mu) / sd, (va - mu) / sd

m_b = train_ae(trs, LATENT)
with torch.no_grad():
    rec_b = m_b(torch.tensor(vas).to(device)).cpu().numpy()

pca_b = PCA(n_components=min(LATENT, len(trs) - 1)).fit(trs.reshape(len(trs), -1))
prec_b = pca_b.inverse_transform(
    pca_b.transform(vas.reshape(len(vas), -1))).reshape(-1, PATCH_T, N_BINS)

ae_ve_b, _, _, _ = call_region_ve(list(rec_b), list(vas))
pca_ve_b, _, _, _ = call_region_ve(list(prec_b), list(vas))
print(f"\n  unseen bat, AE  : {ae_ve_b*100:.1f}% of call variance")
print(f"  unseen bat, PCA : {pca_ve_b*100:.1f}%")
print("  (compare to the leave-one-call-out number — a big drop means the")
print("   code was fitting one individual)")

n = min(3, len(vas))
fig, ax = plt.subplots(3, n, figsize=(3.4 * n, 8), squeeze=False)
for j in range(n):
    for row, (d_, lab, cm) in enumerate([(vas[j], "original (unseen bat)", "inferno"),
                                         (rec_b[j], f"AE ({LATENT})", "inferno"),
                                         (vas[j] - rec_b[j], "residual", "RdBu_r")]):
        a = ax[row][j]
        v = np.abs(d_).max() if cm == "RdBu_r" else None
        a.imshow(d_.T, aspect="auto", origin="lower", cmap=cm, extent=ext,
                 **({"vmin": -v, "vmax": v} if v else {}))
        if j == 0: a.set_ylabel(lab, fontsize=9)
plt.suptitle("Bout 1 — model never saw this bat", y=1.0)
plt.tight_layout(); plt.savefig(f"{OUT}/test_b_bout_transfer.png", dpi=140,
                                bbox_inches="tight"); plt.show()

# =========================================================
# A. Latent sweep
# =========================================================
print("\n" + "=" * 62); print("A. LATENT SWEEP"); print("=" * 62)
rows = []
for d in [2, 3, 4, 6, 8, 12, 16, 24]:
    folds = [run_fold(h, latent=d) for h in range(N)]
    ae_ve, ae_corr, ae_ce, _ = call_region_ve([f["ae"] for f in folds],
                                              [f["orig"] for f in folds])
    pca_ve, _, pca_ce, _ = call_region_ve([f["pca"] for f in folds],
                                          [f["orig"] for f in folds])
    gain = (pca_ce - ae_ce) / pca_ce * 100
    rows.append((d, ae_ve, pca_ve, gain))
    print(f"  latent {d:2d}:  AE {ae_ve*100:5.1f}%   PCA {pca_ve*100:5.1f}%   "
          f"AE-PCA {gain:+5.1f}%")

dd, ae_v, pca_v, gains = zip(*rows)
fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].plot(dd, np.array(ae_v) * 100, "o-", label="autoencoder")
ax[0].plot(dd, np.array(pca_v) * 100, "s--", label="PCA")
ax[0].set_xlabel("latent dims"); ax[0].set_ylabel("% call variance")
ax[0].legend(); ax[0].grid(alpha=.3); ax[0].set_title("look for the elbow")
ax[1].axhline(0, c="k", lw=.8); ax[1].plot(dd, gains, "o-", color="tab:purple")
ax[1].set_xlabel("latent dims"); ax[1].set_ylabel("AE advantage over PCA (%)")
ax[1].grid(alpha=.3); ax[1].set_title("does nonlinearity ever pay?")
plt.tight_layout(); plt.savefig(f"{OUT}/test_a_latent_sweep.png", dpi=140,
                                bbox_inches="tight"); plt.show()

# =========================================================
# C. What do the dimensions mean?
# =========================================================
print("\n" + "=" * 62); print("C. LATENT INTERPRETABILITY"); print("=" * 62)

mu_c = calls.mean(axis=(0, 1), keepdims=True)
sd_c = calls.std(axis=(0, 1), keepdims=True); sd_c[sd_c < 1e-8] = 1.0
Xs = ((calls - mu_c) / sd_c).astype(np.float32)
aug = np.concatenate([Xs + g / sd_c.mean() for g in GAINS_DB], axis=0)
m_c = train_ae(aug, LATENT)
with torch.no_grad():
    Z = m_c.encode(torch.tensor(Xs).to(device)).cpu().numpy()


def measure(patch):
    z = (patch - patch.mean()) / (patch.std() + 1e-9)
    mask = z > 1.5
    if mask.sum() < 5: return None
    frames = np.where(mask.any(axis=1))[0]
    bins = np.where(mask.any(axis=0))[0]
    trace = np.array([BAND[int(z[t].argmax())] / 1000 for t in frames])
    return dict(peak_khz=BAND[int(z.max(axis=0).argmax())] / 1000,
                start_khz=trace[0], end_khz=trace[-1],
                slope=(trace[-1] - trace[0]) / max(len(frames), 1),
                duration_ms=len(frames) * HOP / SR * 1000,
                bandwidth_khz=(BAND[bins.max()] - BAND[bins.min()]) / 1000,
                energy=float(patch[mask].mean()))


params, idx = [], []
for i in range(N):
    p = measure(calls[i])
    if p: params.append(p); idx.append(i)

if len(params) >= 6:
    keys = list(params[0])
    P = np.array([[p[k] for k in keys] for p in params]); Zi = Z[idx]
    print(f"\n  {len(params)} calls measured")
    print(f"  {'parameter':16s}" + "".join(f"  z{i}  " for i in range(LATENT)) + "   R2")
    for j, k in enumerate(keys):
        y = P[:, j]
        if np.std(y) < 1e-9: continue
        cors = [np.corrcoef(Zi[:, d], y)[0, 1] for d in range(LATENT)]
        r2 = LinearRegression().fit(Zi, y).score(Zi, y)
        print(f"  {k:16s}" + "".join(f"{c:+5.2f} " for c in cors) +
              f"   {r2:5.2f}" + ("  <-- encoded" if r2 > 0.5 else ""))

    fig, ax = plt.subplots(1, 3, figsize=(13, 3.8))
    for a, k in zip(ax, keys[:3]):
        y = P[:, keys.index(k)]
        d = int(np.argmax([abs(np.corrcoef(Zi[:, i], y)[0, 1]) for i in range(LATENT)]))
        a.scatter(Zi[:, d], y, s=40); a.grid(alpha=.3)
        a.set_xlabel(f"latent dim {d}"); a.set_ylabel(k)
        a.set_title(f"r = {np.corrcoef(Zi[:, d], y)[0,1]:+.2f}")
    plt.tight_layout(); plt.savefig(f"{OUT}/test_c_latent_meaning.png", dpi=140,
                                    bbox_inches="tight"); plt.show()
else:
    print(f"  only {len(params)} measurable calls — skipping")

# =========================================================
# D. Bit budget
# =========================================================
print("\n" + "=" * 62); print("D. BIT BUDGET"); print("=" * 62)

lo, hi = Z.min(axis=0), Z.max(axis=0)
print(f"\n  {'bits/dim':>9s} {'bytes/call':>11s} {'vs patch':>10s} {'call var':>10s}")
res = []
for bits in QUANT_BITS:
    lv = 2 ** bits - 1
    Zq = np.round((Z - lo) / (hi - lo + 1e-12) * lv) / lv * (hi - lo) + lo
    ve, _, _, _ = call_region_ve(list(decode(m_c, Zq)), list(Xs))
    nb = LATENT * bits / 8
    res.append((bits, nb, ve))
    print(f"  {bits:9d} {nb:11.1f} {PATCH_T*N_BINS/nb:9.0f}x {ve*100:9.1f}%")

q = np.clip((Xs - Xs.min()) / (Xs.max() - Xs.min()) * 255, 0, 255).astype(np.uint8)
zl = np.mean([len(zlib.compress(q[i].tobytes(), 6)) for i in range(N)])
print(f"\n  baselines per call:")
print(f"    raw patch float32   {PATCH_T*N_BINS*4:8.0f} bytes")
print(f"    raw patch int8      {PATCH_T*N_BINS:8.0f} bytes")
print(f"    int8 + zlib         {zl:8.0f} bytes")
print(f"    {LATENT} latents @ 8 bit   {LATENT:8.0f} bytes")
print(f"\n  vs int8+zlib (the honest baseline): {zl/LATENT:.0f}x smaller")
print(f"  1 s raw audio @384kHz/16bit = {384000*2/1024:.0f} KB — detection alone")
print(f"  already provides most of the reduction")

bts, nbs, ves = zip(*res)
plt.figure(figsize=(7, 4))
plt.plot(bts, np.array(ves) * 100, "o-")
plt.xlabel("bits per latent dimension"); plt.ylabel("% call variance")
plt.title("how much precision must actually be sent")
plt.grid(alpha=.3); plt.tight_layout()
plt.savefig(f"{OUT}/test_d_bit_budget.png", dpi=140, bbox_inches="tight"); plt.show()

print("\n" + "=" * 62)
print(f"SUMMARY")
print(f"  leave-one-call-out (same bat siblings in train): see earlier run")
print(f"  unseen bat (bout transfer):  AE {ae_ve_b*100:.1f}%  PCA {pca_ve_b*100:.1f}%")
print(f"  best latent width by call variance: "
      f"{dd[int(np.argmax(ae_v))]} dims at {max(ae_v)*100:.1f}%")
print("=" * 62)
```
